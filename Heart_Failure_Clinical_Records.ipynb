8.**Visualization**

```python
plt.figure(figsize=(8,5))
plt.hist(df["age"], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
df["DEATH_EVENT"].value_counts().plot(kind="bar")
plt.title("Death Cases")
plt.show()
```

```python
df["sex"].value_counts().plot(kind="bar")
plt.title("Gender Distribution")
plt.show()
```

```python
df["diabetes"].value_counts().plot(kind="bar")
plt.title("Diabetes")
plt.show()
```

```python
df["smoking"].value_counts().plot(kind="bar")
plt.title("Smoking")
plt.show()
```

```python
df["high_blood_pressure"].value_counts().plot(kind="bar")
plt.title("High Blood Pressure")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["age"], df["ejection_fraction"])
plt.xlabel("Age")
plt.ylabel("Ejection Fraction")
plt.title("Age vs Ejection Fraction")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["age"], df["serum_creatinine"])
plt.xlabel("Age")
plt.ylabel("Serum Creatinine")
plt.title("Age vs Serum Creatinine")
plt.show()
```

9. **Insights**

* Dataset contains 299 heart
   failure patients.
* Majority of patients are     male.
* Older patients show a higher mortality rate.
* Patients with lower ejection fraction have a higher risk of death.
* High serum creatinine is associated with increased mortality.
* Diabetes and high blood pressure are common among patients.
* Smoking is observed in a smaller portion of patients compared to non-smokers.
